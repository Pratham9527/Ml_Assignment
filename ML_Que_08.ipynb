```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB
from sklearn.metrics import accuracy_score, f1_score, classification_report
```

```python
data = pd.read_csv("pima-indians-diabetes.data.csv")
```

```python
data.columns = ['Pregnancies','Glucose','BloodPressure','SkinThickness',
                'Insulin','BMI','DiabetesPedigree','Age','Outcome']
```

```python
X = data.drop('Outcome', axis=1)
y = data['Outcome']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
gnb = GaussianNB()
gnb.fit(X_train, y_train)
y_pred_gnb = gnb.predict(X_test)
acc_gnb = accuracy_score(y_test, y_pred_gnb)
f1_gnb = f1_score(y_test, y_pred_gnb)

print("\n📊 Gaussian Naive Bayes Results:")
print(f"Accuracy: {acc_gnb:.4f}")
print(f"F1-Score: {f1_gnb:.4f}")
print(classification_report(y_test, y_pred_gnb))
```

```python
mnb = MultinomialNB()
mnb.fit(X_train, y_train)
y_pred_mnb = mnb.predict(X_test)
acc_mnb = accuracy_score(y_test, y_pred_mnb)
f1_mnb = f1_score(y_test, y_pred_mnb)

print("\n📊 Multinomial Naive Bayes Results:")
print(f"Accuracy: {acc_mnb:.4f}")
print(f"F1-Score: {f1_mnb:.4f}")
print(classification_report(y_test, y_pred_mnb))
```

```python
comparison = pd.DataFrame({
    'Model': ['GaussianNB', 'MultinomialNB'],
    'Accuracy': [acc_gnb, acc_mnb],
    'F1-Score': [f1_gnb, f1_mnb]
})

print("\n🧾 Model Comparison Summary:")
print(comparison)
```
